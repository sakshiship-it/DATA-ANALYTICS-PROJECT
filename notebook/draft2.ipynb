```python
import pandas as pd
import numpy as np

df = pd.read_csv("health_data.csv")
df.head()
```

```python
# Creating arrays
a = np.array([10, 20, 30, 40, 50])
print(a, a.dtype, a.shape)

print(np.arange(0, 10, 2))      # 0,2,4,6,8
print(np.zeros(3), np.ones(3))
print(np.linspace(0, 1, 5))
```

```python
# Array maths works on all elements at once
print(a + 5)
print(a * 2)
print("Mean:", a.mean(), " Sum:", a.sum(), " Max:", a.max(), " Std:", a.std())
```

```python
# 2D array and indexing
m = np.array([[1, 2, 3], [4, 5, 6]])
print(m)
print(m[0, 1])      # row 0, column 1
print(m[:, 2])      # all rows, column 2
print(m.T)          # transpose
```

```python
# NumPy on a dataframe column
ages = df["Age"].to_numpy()
print("Mean age  :", np.mean(ages))
print("Median age:", np.median(ages))
print("Min / Max :", ages.min(), ages.max())   # look at these carefully!
```

```python
# Selecting columns
print(df["City"].head())
print(df[["Patient_ID", "Age", "BMI"]].head())
```

```python
# Selecting rows: iloc (by position) and loc (by label/condition)
print(df.iloc[0])
print(df.iloc[0:5, 0:4])
print(df.loc[df["City"] == "Mumbai"].head())
```

```python
# Filtering with conditions
print(df[df["Age"] > 60].shape[0], "patients older than 60")
print(df[(df["Gender"] == "Female") & (df["BMI"] > 30)].shape[0], "females with BMI > 30")
```

```python
# Summary statistics
df.describe()
```

```python
# Counting values in categorical columns
print(df["Gender"].value_counts())
print(df["Admission_Type"].value_counts())
print(df["City"].value_counts())   # notice the messy city names!
```

```python
# Group by: average values per group
print(df.groupby("Admission_Type")["Hospital_Stay_Days"].mean())
print(df.groupby("Gender")[["BMI", "Fasting_Blood_Sugar"]].mean())
```

```python
# Sorting
print(df.sort_values("Hospital_Stay_Days", ascending=False).head())
```

```python
# Creating a new column
df["BMI_Category"] = pd.cut(df["BMI"], bins=[0, 18.5, 25, 30, 100],
                            labels=["Underweight", "Normal", "Overweight", "Obese"])
print(df["BMI_Category"].value_counts())
```

```python
# Using NumPy with pandas: np.where
df["High_Sugar"] = np.where(df["Fasting_Blood_Sugar"] > 125, "Yes", "No")
print(df["High_Sugar"].value_counts())
```

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nSuspicious ages:")
print(df[(df["Age"] < 0) | (df["Age"] > 110)][["Patient_ID", "Age"]])
```
