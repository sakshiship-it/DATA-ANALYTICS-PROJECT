```python
# Step 1: Read the CSV file using pandas
import pandas as pd

df = pd.read_csv("health_data.csv")
print("File loaded successfully!")
```

```python
# First 5 rows and last 5 rows
print(df.head())
print()
print(df.tail())
```

```python
# Shape = (rows, columns)
print("Shape:", df.shape)
print("Number of rows   :", df.shape[0])
print("Number of columns:", df.shape[1])
print("Column names     :", list(df.columns))
```

```python
# Basic info about the columns and data types
df.info()
```

```python
# Read only some columns
small = pd.read_csv("health_data.csv", usecols=["Patient_ID", "City", "Age"])
print(small.head())

# Read only the first 10 rows
first10 = pd.read_csv("health_data.csv", nrows=10)
print(first10.shape)
```

```python
# Write the first 20 rows to a new CSV file (index=False avoids an extra index column)
df.head(20).to_csv("sample_20_rows.csv", index=False)
print("sample_20_rows.csv saved")

# Read it back to confirm
check = pd.read_csv("sample_20_rows.csv")
print(check.shape)
```

```python
# Write to an Excel file (needs openpyxl: run `pip install openpyxl` if you get an error)
df.head(20).to_excel("sample_20_rows.xlsx", index=False)
print("sample_20_rows.xlsx saved")
```

```python
# Plain Python file handling: write and read a text file
with open("notes.txt", "w") as f:
    f.write("Healthcare data project\n")
    f.write("Dataset has " + str(df.shape[0]) + " rows\n")

with open("notes.txt", "r") as f:
    print(f.read())
```
